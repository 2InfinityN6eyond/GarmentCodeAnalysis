```python
# 통합할때 추가하는 방법. (2024.02.04 작성)
import json
import os
def unify_clo_json_data(TOP_GARMENT_PATH, BOTTOM_GARMENT_PATH):
    with open(TOP_GARMENT_PATH, 'r') as file:
        TOP_GARMENT_DATA = json.load(file)
    with open(BOTTOM_GARMENT_PATH, 'r') as file:
        BOTTOM_GARMENT_DATA = json.load(file)
    
    # FabricList, GradingRuleTableList(제껴도됨), PatternList, SymmetricDataList(제껴도됨), InstanceDataList, SeamLinePairGroupList
    # 통합할때 추가하는 방법.
    # Fabric의 경]우
    TOP_GARMENT_DATA['FabricList'].append(BOTTOM_GARMENT_DATA['FabricList'][0])
    # Pattern의 경우
    [TOP_GARMENT_DATA['PatternList'].append(BOTTOM_GARMENT_DATA['PatternList'][num]) for num in range(len(BOTTOM_GARMENT_DATA['PatternList']))]
    # InstanceDataList의 경우
    [TOP_GARMENT_DATA['InstanceDataList'].append(BOTTOM_GARMENT_DATA['InstanceDataList'][num]) for num in range(len(BOTTOM_GARMENT_DATA['InstanceDataList']))]
    # SeamLinePairGroupList의 경우
    [TOP_GARMENT_DATA['SeamLinePairGroupList'].append(BOTTOM_GARMENT_DATA['SeamLinePairGroupList'][num]) for num in range(len(BOTTOM_GARMENT_DATA['SeamLinePairGroupList']))]
    
    return TOP_GARMENT_DATA # TOP에 BOTTOM 추가하는 방식임.

from copy import deepcopy

def combine_top_bottom_clo_json(
    top_clo_json, bottom_clo_json
) :
    combined_clo_json = deepcopy(top_clo_json)
    combined_clo_json["FabricList"].extend(bottom_clo_json["FabricList"])
    combined_clo_json["PatternList"].extend(bottom_clo_json["PatternList"])
    combined_clo_json["InstanceDataList"].extend(bottom_clo_json["InstanceDataList"])
    combined_clo_json["SeamLinePairGroupList"].extend(bottom_clo_json["SeamLinePairGroupList"])
    return combined_clo_json


# def stitch_data_unify( TOP_STITCH_PATH, BOTTOM_STITCH_PATH):

#     with open(TOP_STITCH_PATH, "r") as f:
#         stitch_data_1 = json.load(f)
#     #stitch_data_1
#     with open(BOTTOM_STITCH_PATH, "r") as f:
#         stitch_data_2 = json.load(f)
    
#     max_panel_1 = max(d["panel"] for d in stitch_data_1)
#     offset = max_panel_1 + 1
#     for item in stitch_data_2:
#         item["panel"] += offset
#     garment_first_id = os.path.basename(os.path.dirname(TOP_STITCH_PATH))
#     garment_second_id = os.path.basename(os.path.dirname(BOTTOM_STITCH_PATH))

#     return (stitch_data_1+stitch_data_2), [garment_first_id, garment_second_id]
```

```python
import os
from tqdm import tqdm
from pathlib import Path
import pandas as pd
import json

with open("outfit_path_list.json", "r") as f :
    outfit_path_list = json.load(f)
with open("top_bottom_path_list.json", "r") as f :
    top_bottom_path_list = json.load(f)

print(len(outfit_path_list))
print(len(top_bottom_path_list))

for p1_p2 in tqdm(top_bottom_path_list) :
    try :
        top_base_path, bottom_base_path = p1_p2.split(",")

        top_garment_split, _, top_garment_id = list(Path(top_base_path).parts)[-3:]
        TOP_PATH = os.path.join(
            "SAMPLE_DATA", top_garment_split, top_garment_id,
            f"{top_garment_id}__00__clo.json"
        )

        bottom_garment_split, _, bottom_garment_id = list(Path(bottom_base_path).parts)[-3:]
        BOTTOM_PATH = os.path.join(
            "SAMPLE_DATA", bottom_garment_split, bottom_garment_id,
            f"{bottom_garment_id}__00__clo.json"
        )

        # top_garment_split, top_garment_id = os.path.basename(TOP_PATH).split("__")[:-1]
        # bottom_garment_split, bo
        # ttom_garment_id = os.path.basename(BOTTOM_PATH).split("__")[:-1]

        with open(TOP_PATH, 'r') as file:
            top_json = json.load(file)

        with open(BOTTOM_PATH, 'r') as file:
            bottom_json = json.load(file)

        combined_clo_json = combine_top_bottom_clo_json(top_json, bottom_json)


        COMBINED_CLO_JSON_PATH = os.path.join(
            "SAMPLE_DATA", top_garment_split, 
            f"{top_garment_id}__00__{bottom_garment_id}__00",
            f"{top_garment_id}__00__{bottom_garment_id}__00__clo.json"
        )
        os.makedirs(os.path.dirname(COMBINED_CLO_JSON_PATH), exist_ok=True)
        
        with open(COMBINED_CLO_JSON_PATH, 'w') as file:
            json.dump(combined_clo_json, file, indent=4)
    except Exception as e :
        print(e)
        print(p1_p2)
        print("="*100)
        continue
```
