```python
import os, sys
from glob import glob
import shutil


DATASET_ROOT = "/home/hjp/VTO2025/GarmentCodeData"
GARMENT_ROOT_PATH = os.path.join(DATASET_ROOT, "GarmentCodeData_v2")

garment_path_list = list(filter(
    lambda x : os.path.isdir(x),
    # lambda x : os.path.isfile(x),
    glob(
        os.path.join(GARMENT_ROOT_PATH, "*", "random_body", "*")
    )
))
```

```python
for garment_path in garment_path_list :
    if os.path.isfile(garment_path) :
        print(garment_path)
        continue
    else :
        # remove directory
        shutil.rmtree(garment_path)


```
